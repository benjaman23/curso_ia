### Ejercicio 9 - Detector de saludo
Solicite un mensaje. Si contiene "hola" o "buenas", mostrar "Saludo detectado"; en caso contrario, "Sin saludo".

```python
# Ejercicio 9 - Detector de saludo
mensaje = input("Ingrese un mensaje: ").lower()

if "hola" in mensaje or "buenas" in mensaje:
    print("Saludo detectado")
else:
    print("Sin saludo")
```

### Ejercicio 10 - Clasificador de tickets
Clasifique un mensaje como Soporte, Ventas u Otro utilizando palabras clave definidas por usted.

```python
# Ejercicio 10 - Clasificador de tickets
mensaje = input("Ingrese su consulta o ticket: ").lower()

# Evaluamos primero palabras clave de Soporte y luego de Ventas
if "error" in mensaje or "ayuda" in mensaje or "falla" in mensaje or "problema" in mensaje:
    print("Soporte")
elif "precio" in mensaje or "comprar" in mensaje or "cotizar" in mensaje or "valor" in mensaje:
    print("Ventas")
else:
    print("Otro")
```

### Ejercicio 11 - Contar palabras sospechosas
Recorra una lista de palabras sospechosas y cuente cuántas aparecen en el mensaje.

```python
# Ejercicio 11 - Contar palabras sospechosas
mensaje = input("Ingrese el mensaje a analizar: ").lower()
palabras_sospechosas = ["premio", "gratis", "urgente", "ganaste", "oferta", "sorteo"]

contador = 0
for palabra in palabras_sospechosas:
    if palabra in mensaje:
        contador += 1

print("Cantidad de palabras sospechosas encontradas:", contador)
```

### Ejercicio 12 - Sentimiento básico
Defina palabras positivas y negativas. Recorra ambas listas, calcule un puntaje y clasifique el texto como positivo, neutral o negativo.

```python
# Ejercicio 12 - Sentimiento básico
mensaje = input("Ingrese un comentario o reseña: ").lower()

palabras_positivas = ["bueno", "excelente", "genial", "feliz", "increible", "recomiendo"]
palabras_negativas = ["malo", "pesimo", "terrible", "triste", "horrible", "deficiente"]

puntaje = 0

for palabra in palabras_positivas:
    if palabra in mensaje:
        puntaje += 1

for palabra in palabras_negativas:
    if palabra in mensaje:
        puntaje -= 1

print("Puntaje obtenido:", puntaje)

if puntaje > 0:
    print("Clasificación: Positivo")
elif puntaje < 0:
    print("Clasificación: Negativo")
else:
    print("Clasificación: Neutral")
```

### Ejercicio 13 - Repetir hasta salir
Use `while` para solicitar mensajes repetidamente hasta que el usuario escriba "salir".

```python
# Ejercicio 13 - Repetir hasta salir
mensaje = ""

while mensaje != "salir":
    mensaje = input("Escribe un mensaje (o escribe 'salir' para terminar): ").lower()
    if mensaje != "salir":
        print("Mensaje registrado:", mensaje)

print("Programa finalizado.")
```

### Ejercicio 14 - Palíndromo
Solicite una palabra y determine si se lee igual de izquierda a derecha y de derecha a izquierda.

```python
# Ejercicio 14 - Palíndromo
palabra = input("Ingrese una palabra: ").lower()

# Invertimos la palabra recorriéndola carácter por carácter
palabra_invertida = ""
for letra in palabra:
    palabra_invertida = letra + palabra_invertida

if palabra == palabra_invertida:
    print("Es un palíndromo (se lee igual en ambos sentidos).")
else:
    print("No es un palíndromo.")
```
